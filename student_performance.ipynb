```python
import pandas as pd
import numpy as np

print("Libraries imported successfully")
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Thanu", "Theju", "Chaithu", "Ammu", "Megs",
             "Yashu", "Harshi", "Preethi", "Pavan", "Yatin"],
    "Department": ["CSE", "ECE", "CSE", "ME", "ECE",
                   "CSE", "ME", "ECE", "CSE", "ME"],
    "Python": [85, 67, 92, 45, 76, 58, 34, 81, 73, 62],
    "Sensors": [80, 72, 95, 50, 79, 60, 38, 84, 70, 65],
    "Actuators": [88, 65, 90, 42, 74, 55, 30, 78, 75, 68],
    "Attendance": [95, 88, 98, 75, 90, 82, 60, 93, 85, 80]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Total"] = df["Python"] + df["Sensors"] + df["Actuators"]

print(df[["Name", "Python", "Sensors", "Actuators", "Total"]])
```

```python
df["Average"] = df["Total"] / 3

print(df[["Name", "Total", "Average"]])
```

```python
def find_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 35:
        return "D"
    else:
        return "F"


df["Grade"] = df["Average"].apply(find_grade)

print(df[["Name", "Average", "Grade"]])
```

```python
def check_result(row):
    if row["Python"] >= 35 and row["Sensors"] >= 35 and row["Actuators"] >= 35:
        return "Pass"
    else:
        return "Fail"


df["Result"] = df.apply(check_result, axis=1)

print(df[["Name", "Python", "Sensors", "Actuators", "Result"]])
```

```python
highest = df["Average"].max()
lowest = df["Average"].min()

print("Highest average:", highest)
print("Lowest average:", lowest)
```

```python
print("Top performer:")
print(df[df["Average"] == highest][["Name", "Average"]])

print("Lowest performer:")
print(df[df["Average"] == lowest][["Name", "Average"]])
```

```python
print("Average marks in Python:", df["Python"].mean())
print("Average marks in Sensors:", df["Sensors"].mean())
print("Average marks in Actuators:", df["Actuators"].mean())
```

```python
marks = df["Total"].to_numpy()

print("Average total marks:", np.mean(marks))
print("Highest total marks:", np.max(marks))
print("Lowest total marks:", np.min(marks))
```

```python
top_students = df.sort_values(by="Average", ascending=False)

print("Top 3 Students:")
print(top_students[["Name", "Average", "Grade"]].head(3))
```

```python
df["Total"] = df["Python"] + df["Sensors"] + df["Actuators"]

df["Average"] = df["Total"] / 3

print(df[["Name", "Total", "Average"]])
```

```python
def find_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 35:
        return "D"
    else:
        return "F"


df["Grade"] = df["Average"].apply(find_grade)


def check_result(row):
    if row["Python"] >= 35 and row["Sensors"] >= 35 and row["Actuators"] >= 35:
        return "Pass"
    else:
        return "Fail"


df["Result"] = df.apply(check_result, axis=1)

print(df[["Name", "Total", "Average", "Grade", "Result"]])
```

```python
print("STUDENT PERFORMANCE REPORT")
print("--------------------------")

print(df[["Student_ID", "Name", "Department",
          "Python", "Sensors", "Actuators",
          "Attendance", "Total", "Average",
          "Grade", "Result"]])
```

```python
def check_attendance(attendance):
    if attendance >= 75:
        return "Good"
    else:
        return "Low"


df["Attendance_Status"] = df["Attendance"].apply(check_attendance)

print(df[["Name", "Attendance", "Attendance_Status"]])
```
